Find all movies whose duration is greater than 120 minutes.

```python
# Step 1: Select only movies
movies = df[df['type'] == 'Movie'].copy()

# Handle missing duration values by dropping relevant rows
movies = movies.dropna(subset=['duration'])

# Step 2: Create a numeric duration column
movies['duration_minutes'] = movies['duration'].str.replace(' min', '', regex=False).astype(int)

# Step 3: Filter movies with duration greater than 120 minutes
long_movies = movies[movies['duration_minutes'] > 120]

# Step 4: Display the result
long_movies
```

**Visualization**


What percentage of Netflix's content is Movies vs TV Shows?

```python
df['type'].value_counts().plot(kind='pie',autopct='%1.1f%%')

plt.title("Movies vs TV Shows on Netflix")

plt.show()
```

How has Netflix's content library grown over the years?

```python
import matplotlib.pyplot as plt

df['release_year'].value_counts().sort_index().plot(kind='line')

plt.title("Netflix Content Library Growth")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.grid(True)
plt.show()
```

Which countries contribute the most Netflix content?

```python
df['country'].value_counts().head(10).plot(kind='barh',edgecolor="black",color="yellow")

plt.title("Countries with the Most Netflix Content")
plt.xlabel("Country")
plt.ylabel("Number of Titles")
plt.show()
```

Which genres are most popular on Netflix?

```python
genres = df['listed_in'].str.split(', ').explode()

genres.value_counts().head(10).plot(kind='barh')

plt.title("Top 10 Netflix Genres")
plt.xlabel("Number of Titles")
plt.ylabel("Genre")
plt.show()
```

Which audience group does Netflix mainly target?

```python
df['rating'].value_counts().plot(kind='bar',color="red")
plt.figsize=(2,4)

plt.title("Audience Group Distribution on Netflix")
plt.xlabel("Audience Group")
plt.ylabel("Number of Titles")
plt.show()
```

During which months does Netflix add the most content?

```python
df['month'] = pd.to_datetime(df['date_added'], format='mixed').dt.month

df['month'].value_counts().sort_index().plot(kind='bar',color="purple")

plt.title("Months with the Most Netflix Content")
plt.xlabel("Month")
plt.xticks(rotation=45)
plt.ylabel("Number of Titles")
plt.show()
```

Which directors contribute the most content

```python
df['director'].value_counts().head(10).plot(kind='barh',color="green")

plt.title("Top 10 Directors with the Most Content")
plt.xlabel("Number of Titles")
plt.ylabel("Directors")
plt.show()
```

Which actors appear most frequently?

```python
actors = df['cast'].dropna().str.split(', ').explode()
actors.value_counts().head(10).plot(kind='barh',color="black")

plt.title("Top 10 Actors with the Most Content")
plt.xlabel("Number of Titles")
plt.ylabel("Actors")
plt.show()
```

What is the distribution of movie durations?

```python
# Filter for movies
movies = df[df['type'] == 'Movie'].copy()

# Handle missing duration values by dropping relevant rows
movies = movies.dropna(subset=['duration'])

# Extract numerical duration and convert to int
movies['duration_minutes'] = movies['duration'].str.replace(' min', '', regex=False).astype(int)

# Plot the histogram of movie durations
movies['duration_minutes'].plot(kind='hist', bins=30, edgecolor='black')

plt.title("Distribution of Movie Durations")
plt.xlabel("Duration (minutes)")
plt.ylabel("Number of Movies")
plt.show()
```
